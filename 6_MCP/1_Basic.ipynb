### Let's use MCP in OpenAI Agents SDK  
MCP lets an agent use tools that someone else wrote and runs as a separate program. We start a small client, it spawns the server as a subprocess, and the server reports back the tools it offers. Once you can list one server's tools, you can use any of them the same way.

`async` with starts the server and shuts it down when the block ends. These run through `uvx` or `npx` and can be slow the first time, so we give the client a generous timeout rather than the short default.

```python
# On Windows, a stdio MCP server started from a Jupyter kernel writes to a stderr stream with no
# real file descriptor and crashes with io.UnsupportedOperation: fileno. We send the server's
# stderr to the null device so it always has somewhere real to write, which lets every cell below
# use MCPServerStdio exactly as the OpenAI Agents SDK documents it.

import functools
import subprocess
import agents.mcp.server

agents.mcp.server.stdio_client = functools.partial(agents.mcp.server.stdio_client, errlog=subprocess.DEVNULL)
```

```python
fetch_params = {"command": "uvx", "args": ["--with", "mcp<2","mcp-server-fetch"]}

async with MCPServerStdio(params=fetch_params, client_session_timeout_seconds=60) as server:
    fetch_tools = await server.list_tools()

print(fetch_tools)
```

```python
import textwrap
print(textwrap.fill(f"Tool name: {fetch_tools[0].name}", width=100))
print()
print(textwrap.fill(f"Tool description:\n{fetch_tools[0].description}", width=100))
print()
print(textwrap.fill(f"Input Schema:\n{fetch_tools[0].inputSchema}"))
```

```python
playwright_params = {"command": "npx", "args": [ "@playwright/mcp@latest"]}

async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as server:
    playwright_tools = await server.list_tools()

playwright_tools
```

```python
sandbox_path = os.path.abspath(os.path.join(os.getcwd(), 'sandbox'))
os.makedirs(sandbox_path, exist_ok=True)
```

```python
files_params = {'command': 'npx', 'args': ['-y', '@modelcontextprotocol/server-filesystem', sandbox_path]}

async with MCPServerStdio(params=files_params, client_session_timeout_seconds=60) as server:
    file_tools = await server.list_tools()

file_tools
```

```python
file_tools[0].model_dump()
```

```python
# Agent + tool
INSTRUCTIONS = """
You browse the internet to accomplish your instructions.
Accept cookies and navigate pop-ups as needed.
If one website isn't fruitful, try another.
Be persistent until you have solved your assignment,
trying different options and sites as needed.
When you need to write files, you do that inside the sandbox/ folder only.
"""

TASK = "Find a great recipe for palak paneer on internet, then summarize it in markdown to food.md"

async with MCPServerStdio(params=files_params, client_session_timeout_seconds=60) as mcp_server_files:
    async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as mcp_server_browser:
        agent = Agent(
            name='investigator',
            instructions=INSTRUCTIONS,
            model='gpt-5.4-mini',
            mcp_servers=[mcp_server_files, mcp_server_browser]
        )

        with trace('investigate'):
            result = await Runner.run(agent, TASK, max_turns=20)
            print(result.final_output)
```

```python
INSTRUCTIONS = """
You browse the internet to accomplish your instructions.
Accept cookies and navigate pop-ups as needed.
Use the provided job posting URL as the primary source.
Extract the complete job description, including:
- Job title
- Company name
- Job location
- Job summary
- Key responsibilities
- Required qualifications and experience
- Preferred qualifications
- Required technical skills
- Any other relevant job details

If the page does not display the complete job description, try alternative approaches
such as inspecting the page content, navigating the website, or checking other
official sources for the same job posting.

Do not invent or assume missing information.
Preserve the original meaning of the job requirements.
When you need to write files, write them inside the sandbox/ folder only.
Save the extracted information as a well-structured Markdown file.
"""

TASK = """
Collect the job description from the following Cummins job posting:

https://cummins.jobs/pune-ind/data-scientist-senior/A922F68393BC4AE0B03A5A549E9525DA/job/?vs=1606&utm_source=LinkedIn.com-DE&utm_medium=Social+Media&utm_campaign=LinkedIn.com

Extract the available job details and organize them into a Markdown document.
Save the document to:

sandbox/job_description.md

Include the source URL in the document.
If any information is unavailable, explicitly mention that it was not found.
"""

async with MCPServerStdio(params=files_params, client_session_timeout_seconds=60) as mcp_server_files:
    async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as mcp_server_browser:
        agent = Agent(
            name='job_description_collector',
            instructions=INSTRUCTIONS,
            model='gpt-5.4-mini',
            mcp_servers=[mcp_server_files, mcp_server_browser]
        )

        with trace('job_description_collector'):
            result = await Runner.run(agent, TASK, max_turns=20)
            print(result.final_output)
```
