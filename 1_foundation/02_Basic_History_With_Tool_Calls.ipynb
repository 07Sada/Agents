```python
# Importing libraries and Setup
from dotenv import load_dotenv
import os
from openai import OpenAI
from pypdf import PdfReader
from IPython.display import Markdown, display
import gradio as gr
import json
```

```python
# Loading the API keys
load_dotenv(override=True)
openai = OpenAI()
```

```python
# Load and see the linkedin profile
reader = PdfReader('twin/linkedin_profile.pdf')
linkedin = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        linkedin+=text

display(Markdown(linkedin))
```

```python
# Load and see the resume
reader = PdfReader('twin/resume.pdf')
resume = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        resume+=text

display(Markdown(resume))
```

```python
# Load the summary
with open('twin\summary.txt', 'r', encoding='utf-8') as f:
    summary = f.read()

display(Markdown(summary))
```

```python

system_prompt = f"""
# Your role

You are an AI digital twin representing the person whose website you are on.
You interact with visitors and answer questions about their career, background,
skills, experience, projects, and professional interests.

You represent the person professionally and conversationally.

# About the person

{summary}

# LinkedIn Profile

{linkedin}

# Resume

{resume}

# Rules

- Answer questions related to the person's career, background, skills, experience,
  projects, education, and professional interests.
- Be professional, friendly, concise, and engaging, especially with recruiters,
  potential employers, clients, or professional visitors.
- Stay in character as the person's professional digital twin.
- Use the provided Summary, LinkedIn profile, and Resume as your sources of truth.
- Do not invent or assume information, experience, skills, projects, achievements,
  or technologies.
- If information is not available in the provided context, clearly say:
  "I don't have that information available."
- If the sources contain conflicting information, prefer the most recent or
  specific information and do not guess.
- If asked whether you are the actual person, clearly explain that you are an
  AI digital twin representing them.
- If asked something unrelated to their professional background, politely steer
  the conversation back toward professional topics.

Always prioritize accuracy over completeness. Never make up an answer.
"""
```

```python
# Visualize the system prompt
display(Markdown(system_prompt))
```

```python
# First message to check the LLM response
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Hi - please tell me about yourself"},
]

response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages)
display(Markdown(response.choices[0].message.content))
```

```python
# Function to hold the history in chat
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages)
    return response.choices[0].message.content
```

```python
# Testing the function
chat_response = chat('Please summarize who you are', [])
display(Markdown(chat_response))
```

```python
# Basic Gradio Application
gr.ChatInterface(chat).launch(inbrowser=False)
```

```python
# Basic Tool Function
def record_email_tool(email: str):
    print(f"Tool called to record an email :{email}")
    with open('email.txt', 'a', encoding='utf-8') as f:
        f.write(email + '\n')
    return "Email received"

# Testing the function
record_email_tool('test@test.com')
```

```python
# JSON to explain the tool to LLM
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ['email'],
        "additionalProperties": False
    }
}


tools = [{"type": "function", "function": record_email_tool_json}]
tools
```

```python
# Adding the tools to chat function
def chat(message: str, history: list[str]):
    messages = [{'role': "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages, tools=tools)

    if response.choices[0].finish_reason == 'tool_calls':
        message = response.choices[0].message
        tool_call = message.tool_calls[0]
        # print(tool_call)
        email = json.loads(tool_call.function.arguments).get('email')
        record_email_tool(email)
        messages.append(message)
        messages.append({"role": "tool", "content": "Email Recorded", "tool_call_id": tool_call.id})
        response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages, tools=tools)

    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch(inbrowser=True)
```

```python
# Modifying the chat function to recieve multiple email address
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages, tools=tools)

    while response.choices[0].finish_reason == 'tool_calls':
        message = response.choices[0].message 
        messages.append(message)

        for tool_call in message.tool_calls:
            email = json.loads(tool_call.function.arguments).get('email')
            record_email_tool(email)

            messages.append({"role": "tool", "content": "Email Recorded", "tool_call_id": tool_call.id})
        response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages, tools=tools)
    
    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch(inbrowser=False)
```
