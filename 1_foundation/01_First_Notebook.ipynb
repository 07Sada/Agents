```python
from dotenv import load_dotenv
import os
from openai import OpenAI
from IPython.display import display, Markdown

load_dotenv()
```

```python
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print("API Key loaded!")
```

```python
openai = OpenAI()

messages = [{"role": "user", "content": "What is capital of India?"}]
response = openai.chat.completions.create(model='gpt-4o-mini', messages=messages)

print(f"User message: {messages[0]['content']}")
print(f"LLM response: {response.choices[0].message.content}")
```

```python
# chaining messages 
# First generate a question
question = "Please propose a hard, challengin question to access someone's IQ. Respond only with the question"
messages = [{"role": "user", "content": question}]

display(Markdown(f"**User asked question to LLM**:{messages[0]['content']}"))

# collect the LLM response
llm_question = openai.chat.completions.create(model='gpt-4o-mini', messages=messages)
llm_question = llm_question.choices[0].message.content

display(Markdown(f"**Question given by LLM**:{llm_question}"))

# add the new question in the message format for the LLM
question_message = [{"role": "user", "content": llm_question}]

# collect the LLM answer
llm_answer = openai.chat.completions.create(model='gpt-4o-mini', messages=question_message)
llm_answer = llm_answer.choices[0].message.content
display(Markdown(f"**Answer given by LLM to the question**:{llm_answer}"))
```
